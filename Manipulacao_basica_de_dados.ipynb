### MANIPULAÇÃO BÁSICA DE DADOS

#### CSV

```python
# Criar o nosso primeiro arquivo CSV

# importa o módulo csv
import csv

dados = [

       # primeira linha: cabeçalho
       ["id", "produto", "preco"],

       # Dados dos produtos
       [1, "Notebook", 3500],
       [2, "Mouse", 80]      
]

# Abrir ou criar um arquivo chamado "produtos.csv"
# "w" significa write (escrever)
# newline = "" evita linha em branco
# encoding utf-8 permite trabalhar com acentos e caracteres especiais
with open("produtos.csv", "w", newline= "", encoding = "utf-8") as arquivo:

       # cria um objeto responsável por escrever dados no arquivo csv
       escritor = csv.writer(arquivo)

       # escrever todas as linha da lista "dados" dentro do arquivo
       escritor.writerows(dados)

print("Arquivo criado com sucesso!")
```

```python
# Ler o arquivo CSV criado "produtos.csv"

import csv

# "r" significa read (ler)
with open("produtos.csv", "r", encoding= "utf-8") as arquivo:

       # objeto responsável por ler o arquivo produtos.csv
       leitor = csv.reader(arquivo)

       for linha in leitor:
              print(linha)
```

### JSON

```python
# cria um arquivo json

import json

# cria um dicionário com os dados

pessoa = {
       "nome":"Ana",
       "idade": 25,
       "cidade": "João Pessoa"
}

with open("pessoa.json", "w", encoding= "utf-8") as arquivo:

       #salva um dicionário dentro do arquivo json
       # ensure_ascii manter os caracteres (ã, ç, í, ô)
       # indent organiza o arquivo json com 4 indentações
       json.dump(pessoa, arquivo, ensure_ascii= False, indent= 4)
print("Arquivo JSON criado com sucesso!")
```

```python
# ler um arquivo json

import json

with open("pessoa.json", "r", encoding= "utf-8") as arquivo:
       pessoa = json.load(arquivo)

print(pessoa)
```

### EXCEL

```python
# Criando e salvando um arquivo excel (xlsx)
# pip install openpyxl

import openpyxl

# cria um novo arquivo excel
arquivo = openpyxl.Workbook()

# seeleciona a planilha
planilha = arquivo.active

# Escreve os dados nas células

planilha["A1"] = "Nome"
planilha["B1"] = "Idade"

planilha["A2"] = "Ana"
planilha["B2"] = 25

planilha["A3"] = "Carlos"
planilha["B3"] = 30

arquivo.save("alunos.xlsx")

print("Arquivo EXCEL criado com sucesso!")
```

```python
# Lendo o arquivo excel (xlsx)

import openpyxl

# Abre o arquivo excel
arquivo = openpyxl.load_workbook("alunos.xlsx")

planilha = arquivo.active

for linha in planilha.iter_rows(values_only= True):
       print(linha)
```
