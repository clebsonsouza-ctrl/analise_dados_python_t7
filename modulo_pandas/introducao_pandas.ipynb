Uma API (Interface de Programação de Aplicações) é uma ponte que permite a comunicação direta entre dois sistemas diferentes.

No seu exemplo da API do Banco Central:

• O seu código faz um pedido (requisição) enviando uma URL com parâmetros (como as datas desejadas).

• O servidor do Banco Central processa o pedido e devolve uma resposta automática com os dados brutos estruturados (em formato json).

```python
import requests
import pandas as pd

dataInicial = '01/01/2026'
dataFinal = '31/12/2026'

# 188 – INPC
# 189 – IGP‑M
# 190 – IGP‑DI
# 433 - IPCA
# 7478 – IPCA‑15
# 10764 – IPCA‑E

url = f'https://api.bcb.gov.br/dados/serie/bcdata.sgs.433/dados?formato=json&dataInicial={dataInicial}&dataFinal={dataFinal}'

response = requests.get(url)
dados = response.json()

df_ipca = pd.DataFrame(dados)

df_ipca
```

### groupby, merge e pivot

#### groupby() — agrupando dados

```python
# Quanto foi vendido em cada município?

df_vendas.groupby("municipio")["valor_liquido"].sum()

# .sum()
# .mean()
# .min()
# .max()
# .count()
```

```python
# Quantas unidades de cada produto foram vendidas?

df_vendas.groupby("produto")["quantidade"].sum()
```

```python
df_faturamento_municipio = (df_vendas.groupby("municipio")["valor_liquido"].sum()
                            .reset_index()
                            .sort_values("valor_liquido", ascending = False))

df_faturamento_municipio
```

```python
# Agrupando por duas colunas

df_vendas.groupby(["municipio", "categoria"])["valor_liquido"].sum().reset_index()
```

```python
# várias métricas ao mesmo tempo

df_resumo = (
    df_vendas
    .groupby("municipio").agg(
        faturamento=("valor_liquido", "sum"),
        quantidade_vendida=("quantidade", "sum"),
        venda_media=("valor_liquido", "mean"),
        numero_vendas=("id_venda", "count")
    ).reset_index()
)

df_resumo
```

### merge() — relacionando duas ou mais tabelas

```python
df_produtos = pd.read_excel("perfil_produtos.xlsx")

df_produtos
```

```python
df_vendas_completo = pd.merge(df_vendas, df_produtos, on="id_produto")

df_vendas_completo
```

```python
df_vendas_completo = pd.merge(df_vendas, df_produtos, left_on= "id_produto", right_on= "id_produto", how = "inner")

# inner - Retorna apenas as linhas que possuem chaves correspondentes em ambas as tabelas.
# left - Mantém todas as linhas da tabela da esquerda e adiciona os dados correspondentes da direita (preenche com NaN se não houver par)
# right - Mantém todas as linhas da tabela da direita e adiciona os dados correspondentes da esquerda (preenche com NaN se não houver par).
# outer - Retorna todas as linhas de ambas as tabelas, combinando onde há correspondência e preenchendo com NaN onde não há.
```

```python
df_fornecedores = pd.read_excel("fornecedores.xlsx")

df_fornecedores
```

```python
df_completo = (

    pd.merge(df_vendas, df_produtos, on = "id_produto", how = 'left')
    .merge(df_fornecedores, on = "fornecedor", how = "left"))

df_completo

# how="left" preservam a tabela que está à esquerda de cada merge.
```

#### pivot_table - cruzar categorias em linhas e colunas

```python
# Faturamento por município e categoria.

df_pivot = pd.pivot_table(
    df_vendas,
    values="valor_liquido", # É o valor que queremos analisar.
    index="municipio", # Serão as linhas.
    columns="categoria", # Serão as colunas.
    aggfunc="sum" # Será a operação realizada.
)

df_pivot
```

```python
# Melhorando a tabela com fill_value

df_pivot = pd.pivot_table(
    df_vendas,
    values="valor_liquido",
    index="municipio",
    columns="categoria",
    aggfunc="sum",
    fill_value=0
).reset_index()

df_pivot
```

```python
# Adicionando totais

df_pivot = pd.pivot_table(
    df_vendas,
    values="valor_liquido",
    index="municipio",
    columns="categoria",
    aggfunc="sum",
    fill_value=0,
    margins=True,
    #margins_name = "Total"
)

df_pivot
```

```python
# Outra Pivot: vendedor × canal

df_pivot = pd.pivot_table(
    df_vendas,
    values="valor_liquido",
    index="vendedor",
    columns="canal_venda",
    aggfunc="sum",
    fill_value=0
)

df_pivot
```
