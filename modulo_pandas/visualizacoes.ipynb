## **Visualização de Dados com Matplotlib e Seaborn**

#### Gráficos básicos (linha, barra, dispersão)

```python
import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns # pip install seaborn
import os
```

```python
# Criando um DataFrame
df = pd.DataFrame({
    "dia": ["Seg", "Ter", "Qua", "Qui", "Sex"],
    "producao": [100, 150, 120, 180, 200]
})

display(df)
```

```python
plt.plot(df["dia"], df["producao"])    # Cria um gráfico de linhas com os dias e a produção
plt.show()                             # Exibe o gráfico na tela
```

```python
# as vezes não colocamos rótulos nas linhas

plt.plot(df["dia"], df["producao"])  # Cria o gráfico de linhas

for dia, producao in zip(df["dia"], df["producao"]):  # Percorre os valores
    plt.text(dia, producao, str(producao))  # Adiciona cada rótulo

plt.show()  # Exibe o gráfico
```

```python
# Personalizando
# Espessura da linha, tamanho dos marcadores e dimensões do gráfico

plt.figure(figsize=(8, 4))           # Define o tamanho da figura: 8 polegadas de largura e 4 de altura

plt.plot(
    df["dia"],                        # Define os valores do eixo X
    df["producao"],                   # Define os valores do eixo Y
    color="red",                      # Define a cor da linha
    marker="o",                       # Utiliza marcadores circulares
    linestyle="-", #linestyle="--"    # Define uma linha contínua
    linewidth=1,                      # Define a espessura da linha
    markersize=8                      # Define o tamanho dos marcadores
)                                    

# for dia, producao in zip(df["dia"], df["producao"]):  # Percorre os valores
#     plt.text(dia, producao, str(producao))  # Adiciona cada rótulo

plt.title("Produção Semanal")         # Adiciona o título

plt.xlabel("Dia da Semana")           # Define o nome do eixo X

plt.ylabel("Quantidade Produzida")    # Define o nome do eixo Y

plt.show()                            # Exibe o gráfico
```

```python
# Alterando a cor de fundo do gráfico

plt.figure(figsize=(8, 4),
            facecolor="lightblue")         # Cria a figura com fundo azul-claro

plt.gca().set_facecolor("lightgray")       # Altera a cor de fundo da área interna do gráfico

plt.plot(
    df["dia"],                             # Define os valores do eixo X
    df["producao"],                        # Define os valores do eixo Y
    color="blue",                          # Define a cor da linha
    marker="o",                            # Utiliza marcadores circulares
    linewidth=2                            # Define a espessura da linha
)                                          

plt.title("Produção Semanal")               # Adiciona o título ao gráfico
plt.xlabel("Dia da Semana")                 # Define o nome do eixo X
plt.ylabel("Quantidade Produzida")          # Define o nome do eixo Y

plt.show()                                 # Exibe o gráfico
```

Matplotlib — Principais opções de cores

No Matplotlib, podemos personalizar as cores de três elementos principais:

| Elemento | Comando | Finalidade |
|---|---|---|
| Linha | `color="blue"` | Define a cor da linha do gráfico. |
| Área interna | `plt.gca().set_facecolor("lightgray")` | Define a cor de fundo da área onde os dados são apresentados. |
| Área externa | `plt.figure(facecolor="lightblue")` | Define a cor de fundo da figura. |

```python
# Criando gráficos de barras agrupadas

import numpy as np                                  # Importa a biblioteca NumPy

x = np.arange(len(df))                              # Cria as posições numéricas das categorias: 0, 1, 2, 3, 4

largura = 0.35                                      # Define a largura de cada barra

plt.figure(figsize=(9, 5))                         # Define o tamanho da figura

barras1 = plt.bar(x - largura/2, df["producao"], largura, label="Produção", color="steelblue")  # Cria as barras da produção

barras2 = plt.bar(x + largura/2, df["meta"], largura, label="Meta", color="orange")             # Cria as barras das metas

plt.bar_label(barras1, padding=3)                   # Adiciona os valores nas barras da produção

plt.bar_label(barras2, padding=3)                   # Adiciona os valores nas barras das metas

plt.xticks(x, df["dia"])                            # Substitui as posições numéricas pelos dias da semana

plt.title("Produção Realizada x Meta")               # Define o título do gráfico

plt.xlabel("Dia da Semana")                         # Define o nome do eixo X

plt.ylabel("Quantidade Produzida")                  # Define o nome do eixo Y

plt.legend()                                        # Exibe a legenda das duas séries

plt.show()                                          # Exibe o gráfico
```

Agrupamento de barras

| Comando | Finalidade |
|---|---|
| `x = np.arange(len(df))` | Define as posições dos grupos. |
| `largura = 0.35` | Define a largura das barras. |
| `x - largura/2` | Desloca as barras para a esquerda. |
| `x + largura/2` | Desloca as barras para a direita. |
| `plt.xticks(x, df["dia"])` | Identifica os dias no eixo X. |

**Resumo:** O deslocamento `largura/2` permite posicionar duas barras lado a lado, facilitando a comparação dos valores.

Principais diferenças no código — Matplotlib x Seaborn

| Recurso | Matplotlib (`plt`) | Seaborn (`sns`) |
|---|---|---|
| **Importação** | `import matplotlib.pyplot as plt` | `import seaborn as sns` |
| **Gráfico de linhas** | `plt.plot(df["x"], df["y"])` | `sns.lineplot(data=df, x="x", y="y")` |
| **Gráfico de barras** | `plt.bar(df["x"], df["y"])` | `sns.barplot(data=df, x="x", y="y")` |
| **Gráfico de dispersão** | `plt.scatter(df["x"], df["y"])` | `sns.scatterplot(data=df, x="x", y="y")` |
| **Linha de regressão** | Exige cálculo adicional da regressão. | `sns.regplot(data=df, x="x", y="y")` |
| **Agrupamento por cores** | Exige separar ou configurar as séries. | `sns.scatterplot(data=df, x="x", y="y", hue="grupo")` |
| **Cor** | `color="blue"` | `color="blue"` |
| **Tamanho dos pontos** | `s=100` | `s=100` |
| **Transparência** | `alpha=0.5` | `alpha=0.5` |
| **Título** | `plt.title("Título")` | `plt.title("Título")` |
| **Eixos** | `plt.xlabel()` e `plt.ylabel()` | `plt.xlabel()` e `plt.ylabel()` |
| **Subplots** | `fig, ax = plt.subplots()` | `sns.scatterplot(..., ax=ax)` |
| **Estilo visual** | `plt.style.use("ggplot")` | `sns.set_theme(style="whitegrid")` |

Diferença fundamental

**Matplotlib:** informa diretamente os valores das colunas.

`plt.scatter(df["x"], df["y"])`

**Seaborn:** informa o DataFrame e os nomes das colunas.

`sns.scatterplot(data=df, x="x", y="y")`

**Observações:**
- O `sns.barplot()` calcula uma estatística por categoria (média, por padrão), enquanto `plt.bar()` utiliza diretamente as alturas informadas.
- O Seaborn utiliza recursos do Matplotlib para títulos, eixos, layouts e exportação.
- As duas bibliotecas podem ser utilizadas em conjunto.

Subplots e Layouts

```python
df_industria = pd.read_excel("df_industria.xlsx")  # Importa a planilha Excel

# Agrupa os dados por data e soma as peças produzidas
producao_data = df_industria.groupby("data_producao")["pecas_produzidas"].sum().reset_index()

# Agrupa os dados por máquina e soma as peças produzidas
producao_maquina = df_industria.groupby("maquina")["pecas_produzidas"].sum().reset_index()
```

```python
# Criando os dois gráficos

# Subplots utilizando apenas Matplotlib

# Cria uma figura com dois gráficos lado a lado
fig, eixos = plt.subplots(1, 2, figsize=(14, 4))

# GRÁFICO 1 — Linhas
eixos[0].plot(
    producao_data["data_producao"],        # Define os valores do eixo X
    producao_data["pecas_produzidas"],     # Define os valores do eixo Y
    marker="o",                            # Adiciona marcadores circulares
    color="blue"                           # Define a cor da linha
)                                          # Finaliza a criação do gráfico

eixos[0].set_title("Produção por Data")     # Define o título
eixos[0].set_xlabel("Data de Produção")     # Identifica o eixo X
eixos[0].set_ylabel("Peças Produzidas")     # Identifica o eixo Y
eixos[0].tick_params(axis="x", rotation=45) # Rotaciona os rótulos do eixo X

# GRÁFICO 2 — Barras
barras = eixos[1].bar(
    producao_maquina["maquina"],            # Define as categorias do eixo X
    producao_maquina["pecas_produzidas"],   # Define a altura das barras
    color="orange"                          # Define a cor das barras
)                                           # Finaliza a criação do gráfico

eixos[1].bar_label(barras, padding=3, fmt="%.0f")  # Adiciona os valores acima das barras

eixos[1].set_title("Produção por Máquina")  # Define o título
eixos[1].set_xlabel("Máquina")              # Identifica o eixo X
eixos[1].set_ylabel("Peças Produzidas")     # Identifica o eixo Y
eixos[1].tick_params(axis="x", rotation=45) # Rotaciona os rótulos do eixo X

plt.tight_layout()                          # Ajusta os espaços entre os gráficos

plt.show()                                  # Exibe os gráficos
```

```python
# Criando os quatro gráficos

df_industria = pd.read_excel("df_industria.xlsx")  # Importa a planilha

df_industria["data_producao"] = pd.to_datetime(df_industria["data_producao"])  # Converte a coluna para datetime

producao_data = df_industria.groupby("data_producao", as_index=False)["pecas_produzidas"].sum()  # Soma a produção por data

producao_data = producao_data.sort_values("data_producao")  # Organiza as datas em ordem crescente

producao_data["dia"] = producao_data["data_producao"].dt.strftime("%d/%m")  # Formata as datas para dia/mês

producao_maquina = df_industria.groupby("maquina", as_index=False)["pecas_produzidas"].sum()  # Soma a produção por máquina
```

```python
# Cria uma figura com 2 linhas e 2 colunas
fig, eixos = plt.subplots(2, 2, figsize=(12, 8), layout="constrained")

# GRÁFICO 1 — Matplotlib: Produção por data
eixos[0, 0].plot(producao_data["dia"], producao_data["pecas_produzidas"], marker="o", color="steelblue")  # Cria a linha

eixos[0, 0].set_title("Produção por Data")  # Define o título
eixos[0, 0].set_xlabel("Data")  # Identifica o eixo X
eixos[0, 0].set_ylabel("Peças Produzidas")  # Identifica o eixo Y
eixos[0, 0].tick_params(axis="x", rotation=45)  # Rotaciona as datas

# GRÁFICO 2 — Seaborn: Produção por máquina
sns.barplot(data=producao_maquina, x="maquina", y="pecas_produzidas", ax=eixos[0, 1], color="orange")  # Cria as barras

eixos[0, 1].set_title("Produção por Máquina")  # Define o título
eixos[0, 1].set_xlabel("Máquina")  # Identifica o eixo X
eixos[0, 1].set_ylabel("Peças Produzidas")  # Identifica o eixo Y

# GRÁFICO 3 — Matplotlib: Temperatura x Defeitos
eixos[1, 0].scatter(df_industria["temperatura_c"], df_industria["pecas_defeituosas"], color="steelblue", alpha=0.7)  # Cria a dispersão

eixos[1, 0].set_title("Temperatura x Defeitos")  # Define o título
eixos[1, 0].set_xlabel("Temperatura (°C)")  # Identifica o eixo X
eixos[1, 0].set_ylabel("Peças Defeituosas")  # Identifica o eixo Y

# GRÁFICO 4 — Seaborn: Tempo de ciclo x Energia
sns.regplot(data=df_industria, x="tempo_ciclo_seg", y="consumo_energia_kwh", ax=eixos[1, 1], ci=None, color="purple")  # Cria a dispersão e a regressão

eixos[1, 1].set_title("Tempo de Ciclo x Energia")  # Define o título
eixos[1, 1].set_xlabel("Tempo de Ciclo (s)")  # Identifica o eixo X
eixos[1, 1].set_ylabel("Consumo (kWh)")  # Identifica o eixo Y

fig.suptitle("Painel de Indicadores Industriais", fontsize=16)  # Adiciona o título geral da figura

plt.show()  # Exibe os quatro gráficos
```
