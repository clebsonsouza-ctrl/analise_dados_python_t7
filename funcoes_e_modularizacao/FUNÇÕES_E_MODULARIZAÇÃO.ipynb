```python
# Exemplo com filtragem de dados

# Queremos somente vendas acima de determinado valor:

def filtrar_vendas(vendas, minimo):
    resultado = []

    for venda in vendas:
        if venda >= minimo:
            resultado.append(venda) # adicionando valores a lista

    return resultado
```

```python
vendas = [1200, 500, 2300, 800, 3100]

vendas_filtradas = filtrar_vendas(vendas, 1000)

print(vendas_filtradas)
```

#### *args permite que uma função receba uma quantidade variável de argumentos posicionais.

```python
def somar(*args):
    print(args)
```

```python
somar(10, 20, 30, 40, 50)
```

```python
# imagine que queremos calcular a média de várias vendas

def calcular_media(*vendas):

    media = sum(vendas) / len(vendas)

    return media
```

```python
media = calcular_media(
    1200,
    1500,
    1800,
    2100,
    500
)

print(media)

#*args recebe vários argumentos posicionais e os reúne em uma tupla.
#o nome args é apenas uma convenção.
```

#### **kwargs - permite receber uma quantidade variável de argumentos nomeados.

```python
def mostrar_dados(**kwargs):
    print(kwargs)
```

```python
mostrar_dados(
    nome = "Maria",
    peso = 65,
    altura = 1.68
)

# Aqui os dados são armazenados em um dicionário.
# A documentação oficial do Python especifica
# que um parâmetro **nome recebe os argumentos
# nomeados adicionais em forma de dicionário.
```

```python
def mostrar_registro(**dados):

    for chave, valor in dados.items():
        print(chave, ":", valor)
```

```python
mostrar_registro(
    nome = "Maria da Silva",
    idade = 35,
    cidade = "Recife"
)
```

#### Função lambda

```python
# Uma função tradicional pode ser escrita assim

def dobrar(valor):
    return valor * 2

resultado = dobrar(10)

print(resultado)
```

```python
dobrar = lambda valor: valor * 2

print(dobrar(10))
```

```python
#acrescentar 10%
aumentar = lambda valor: valor * 1.10

print(aumentar(1000))
```

#### Exemplo lambda com tabelas 

```python
import pandas as pd

dados = {
    "produto": ["Notebook", "Monitor", "Mouse"],
    "preco": [3500, 1200, 150],
    "quantidade": [2, 3, 10]
}

df = pd.DataFrame(dados)

df
```

```python
df["preco_novo"] = df["preco"].apply(lambda valor: valor * 1.10)

display(df)
```

```python
df["faturamento"] = df.apply(
    lambda linha: linha["preco"] * linha["quantidade"], axis = 1)

df
```

```python
dados = {
    "produto": [
        "  notebook ",
        "MONITOR   ",
        "   mouse"
    ],
    "preco": [3500, 1200, 150]
}

df = pd.DataFrame(dados)

df
```

```python
df["produto"] = df["produto"].apply(lambda texto:
                                     texto.strip().upper())

df
```

```python
# Lambda com uma condição

df["categoria"] = (df["preco"]
                   .apply(lambda valor:
                     "ALTO" if valor >= 1000 else "BAIXO"))

df
```

```python
# lambda com mais de um parâmetro

calcular_total = lambda preco, quantidade: preco * quantidade

resultado = calcular_total(3500, 2)

print(resultado)
```

#### Unindo lambda + tratamento de dados 

```python
dados = {
    "nome": [
        "   maria     da silva ",
        "joão    pereira   ",
        "   ana    souza"
    ],
    "peso": [65, 82, 58],
    "altura": [1.68, 1.80, 1.62]
}

df = pd.DataFrame(dados)

df
```

```python
import re

df["nome"] = df["nome"].apply(
    lambda nome: re.sub(r"\s+", " ", nome).strip().upper()
)

df
```
