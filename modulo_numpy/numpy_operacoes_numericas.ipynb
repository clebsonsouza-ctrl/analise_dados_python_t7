#### Outras formas de criar Arrays

| Função          | Finalidade                                         |
|-----------------|----------------------------------------------------|
| `np.array()`    | Criar array a partir de valores existentes         |
| `np.zeros()`    | Criar array preenchido com zeros                   |
| `np.ones()`     | Criar array preenchido com uns                     |
| `np.arange()`   | Criar sequência numérica com intervalo definido    |
| `np.linspace()` | Criar sequência com valores igualmente espaçados   |

```python
# np.zeros() — Criar um array preenchido com zeros

valores = np.zeros((2, 3))

print(valores)
```

```python
# np.ones() — Criar um array preenchido com uns

valores = np.ones((2, 3))

print(valores)
```

```python
# np.arange() — Criar uma sequência numérica

numeros = np.arange(1, 6)

print(numeros)
```

```python
# np.linspace() — Criar valores igualmente espaçados

""" 
permite determinar quantos valores serão distribuídos entre um número inicial e um número final.

0: valor inicial;
10: valor final, incluído por padrão;
5: quantidade de valores que serão gerados.

"""

numeros = np.linspace(0, 10, 5)

print(numeros)
```

#### NumPy: Operações Vetorizadas

```python
vendas = np.array([100, 200, 300, 400, 500])
```

```python
# Desejamos simular um aumento de 10% em todas as vendas.

novas_vendas = vendas * 1.10

print(novas_vendas)
```

```python
# Uma loja deseja aplicar um desconto de 20% em diferentes produtos.

precos = np.array([100, 200, 300, 400])

precos_desconto = precos * (1 - 0.2)

print(precos_desconto)
```

```python
# Desejamos calcular o faturamento de cada produto.

quantidades = np.array([10, 20, 30])

precos = np.array([5, 8, 12])

faturamento = quantidades * precos

print(faturamento)
```

```python
# Operações vetorizadas em arrays bidimensionais (2D)

vendas = np.array([
    [100, 200, 300],  # Loja 1
    [150, 250, 350]   # Loja 2
])

# A empresa deseja simular um aumento de 10% nas vendas de todas as lojas.

novas_vendas = vendas * 1.10

print(novas_vendas)
```

```python
# Filtragem de arrays com condições

# A empresa estabeleceu uma meta diária de R$ 200.

vendas = np.array([120, 250, 180, 90, 310, 200])

resultado = vendas > 200

print(resultado)
```

```python
resultado = vendas[vendas > 200]

print(resultado)
```

```python
# Imagine que queremos selecionar vendas entre R$ 150 e R$ 250, incluindo os dois limites.

resultado = vendas[(vendas >= 150) & (vendas <= 250)]

print(resultado)
```

| Operador | Significado         |
|----------|---------------------|
| `&`      | E lógico (AND)      |
| `\|`     | OU lógico (OR)      |
| `~`      | Negação lógica (NOT) |

```python
# Operações condicionais com np.where()

#np.where(condicao, valor_verdadeiro, valor_falso)

""" 
A empresa deseja classificar as vendas como:
- "Meta atingida": vendas maiores ou iguais a R$ 200.
- "Abaixo da meta": vendas inferiores a R$ 200.
"""

situacao = np.where(vendas >= 200, "Meta atingida", "Abaixo da meta")

print(situacao)
print(30*"-")
print(vendas)
```

```python
# Calculando bônus de vendas

bonus = np.where(vendas >= 200, vendas * 0.10, 0)

print(bonus)
print(30*"-")
print(vendas)
```

### Principais funções estatísticas

| Função        | Finalidade          |
|---------------|---------------------|
| `np.sum()`    | Soma dos elementos  |
| `np.mean()`   | Média aritmética    |
| `np.median()` | Mediana             |
| `np.min()`    | Menor valor         |
| `np.max()`    | Maior valor         |
| `np.std()`    | Desvio padrão       |

```python
# Funções estatísticas com NumPy
print(vendas)
print(30*"*")
print("Soma vendas:", np.sum(vendas))
print("Média das vendas:", np.mean(vendas))
print("Mediana das vendas:", np.median(vendas))
print("Menor venda", np.min(vendas))
print("Menor venda", np.max(vendas))
```

```python
# Arredondando resultados com np.round()

media = np.mean(vendas)

media_arredondada = np.round(media, 2)

print(media_arredondada)
```

#### NumPy: Broadcasting

##### Uma empresa possui duas lojas e comercializa três produtos.

| Loja   | Produto A | Produto B | Produto C |
|--------|----------:|----------:|----------:|
| Loja 1 | 100       | 200       | 300       |
| Loja 2 | 150       | 250       | 350       |


A empresa deseja aplicar diferentes reajustes:
- Produto A: aumento de 10%;
- Produto B: aumento de 20%;
- Produto C: aumento de 5%.

```python
# Broadcasting entre array 2D e array 1D

precos = np.array([
    [100, 200, 300], # loja 01
    [150, 250, 350] # loja 02
])

reajustes = np.array([1.10, 1.20, 1.05])
```

```python
# Aplicando o Broadcasting

novos_precos = precos * reajustes

print(novos_precos)
```

##### Reajuste por produto aplicado às duas lojas

| Loja / Reajuste | Produto A | Produto B | Produto C |
|-----------------|----------:|----------:|----------:|
| **Reajuste**    | × 1,10    | × 1,20    | × 1,05    |
| Loja 1          | 100       | 200       | 300       |
| Loja 2          | 150       | 250       | 350       |

O que aconteceu?

O NumPy aplicou os reajustes às colunas correspondentes:

- Primeira coluna multiplicada por 1.10;
- Segunda coluna multiplicada por 1.20;
- Terceira coluna multiplicada por 1.05.

```python
# Operação entre matrizes de mesmo formato

matriz_a = np.array([
    [10, 20, 30],
    [40, 50, 60]
])

matriz_b = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

resultado = matriz_a + matriz_b

print(resultado)
```

```python
# Broadcasting por linha

vendas = np.array([
    [100, 200, 300],
    [150, 250, 350]
])

bonus = np.array([
    [10],
    [20]
])

resultado = vendas + bonus

print(resultado)
```

##### Broadcasting por linha

| Loja   | Produto A | Produto B | Produto C |
|--------|-----------|-----------|-----------|
| Loja 1 | 100 + 10 = **110** | 200 + 10 = **210** | 300 + 10 = **310** |
| Loja 2 | 150 + 20 = **170** | 250 + 20 = **270** | 350 + 20 = **370** |

> #### ⚠️ **Atenção:** Nem todos os arrays podem ser combinados.


No Broadcasting, comparamos os tamanhos dos eixos da direita para a esquerda.
Cada par deve ter tamanhos iguais OU pelo menos um tamanho igual a 1.

Regras de compatibilidade do Broadcasting

| Array A  | Array B  | Compatível? | Resultado |
|----------|----------|-------------|-----------|
| `(3,)`   | `(3,)`   | Sim         | `(3,)`    |
| `(2, 3)` | `(3,)`   | Sim         | `(2, 3)`  |
| `(2, 3)` | `(1, 3)` | Sim         | `(2, 3)`  |
| `(2, 3)` | `(2, 1)` | Sim         | `(2, 3)`  |
| `(2, 3)` | `(2,)`   | Não         | Erro      |
| `(2, 3)` | `(4,)`   | Não         | Erro      |
