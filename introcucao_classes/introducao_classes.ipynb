## **INTRODUÇAO A CLASSES**

Exemplo simples de classe

```python
# Define a classe Maquina
class Maquina:

    # Define um método para apresentar a máquina
    def apresentar(self, nome):
        print("Máquina:", nome)  # Exibe o nome recebido


# Cria um objeto da classe Maquina
maquina1 = Maquina()

# Executa o método, informando o nome
maquina1.apresentar("M01")

# O método apresentar() recebeu o nome "M01" como argumento e utilizou essa informação para exibir uma mensagem.
# Entretanto, o nome não foi armazenado como atributo do objeto.
```

Agora imagine que essa máquina tenha vários métodos, como ligar, desligar e consultar a produção.

def __init__(self, nome):

O __init__ é um método especial de inicialização,

executado automaticamente quando uma nova instância da classe é criada.

```python
# Define a classe Maquina
class Maquina:

    # Define o método construtor
    def __init__(self, nome):
        self.nome = nome  # Armazena o nome da máquina

        # self permite que cada objeto acesse seus próprios atributos e métodos.
        # nome: parâmetro que recebe temporariamente o valor "M01".
        # self.nome: atributo que guarda o valor dentro daquele objeto.

    # Define um método para apresentar a máquina
    def apresentar(self):
        print("Máquina:", self.nome)  # Exibe o nome armazenado


# Cria um objeto informando seu nome
maquina1 = Maquina("M01")

# Executa o método apresentar
maquina1.apresentar()
```

| Elemento | Significado |
|---|---|
| `def` | Define um método. |
| `__init__` | Método especial utilizado para inicializar o objeto. |
| `self` | Referência ao objeto que está sendo inicializado. |
| `nome` | Parâmetro que recebe o nome informado na criação do objeto. |

```python
# Define a classe Maquina
class Maquina:

    # Inicializa a máquina com seu nome
    def __init__(self, nome):
        self.nome = nome  # Armazena o nome no objeto

    # Define um método para ligar a máquina
    def ligar(self):
        print(self.nome, "foi ligada!")  # Utiliza o nome armazenado

    # Define um método para desligar a máquina
    def desligar(self):
        print(self.nome, "foi desligada!")  # Reutiliza o mesmo nome


# Cria um objeto chamado maquina1
maquina1 = Maquina("M01")

# Os métodos ligar() e desligar() conseguem utilizar o nome porque ele está armazenado no atributo self.nome.
```

```python
# Executa os métodos
maquina1.ligar()  # Liga a máquina
```

```python
maquina1.desligar()  # Desliga a máquina
```

| Sem `__init__` | Com `__init__` |
|---|---|
| `maquina1 = Maquina()` | `maquina1 = Maquina("M01")` |
| `maquina1.apresentar("M01")` | `maquina1.apresentar()` |
| O nome é passado ao método. | O nome já está armazenado no objeto. |
| Ideal para operações que não precisam guardar o nome. | Útil quando vários métodos precisam acessar o nome. |

Observação:

para simplesmente imprimir "M01",
uma função tradicional seria suficiente.

Não precisamos criar classes para toda operação em Python.

O __init__ se torna útil quando desejamos inicializar

informações relacionadas a um objeto para utilizá-las posteriormente.

Exemplo de classe com livros

```python
# Define a classe Livro
class Livro:

    # Inicializa os atributos do livro
    def __init__(self, titulo, autor, preco):
        self.titulo = titulo  # Armazena o título
        self.autor = autor    # Armazena o autor
        self.preco = preco    # Armazena o preço

    # Método para apresentar as informações
    def apresentar(self):
        print("Título:", self.titulo)  # Exibe o título
        print("Autor:", self.autor)    # Exibe o autor
        print("Preço: R$", self.preco)  # Exibe o preço
```

```python
# Cria dois objetos da classe Livro
livro1 = Livro("Dom Casmurro", "Machado de Assis", 39.90)
livro2 = Livro("O Cortiço", "Aluísio Azevedo", 29.90)
#Conceito principal: cada livro é um objeto independente,
#mas ambos compartilham a mesma estrutura definida pela classe Livro.
```

```python
# Executa o método apresentar para cada objeto
livro1.apresentar()
```

```python
# Executa o método apresentar para cada objeto
livro2.apresentar()
```

| Elemento | Finalidade |
|---|---|
| `class Livro:` | Define a classe. |
| `__init__()` | Inicializa os atributos do livro. |
| `self.titulo` | Armazena o título no objeto. |
| `self.autor` | Armazena o autor no objeto. |
| `self.preco` | Armazena o preço no objeto. |
| `apresentar()` | Exibe as informações do livro. |
| `livro1` e `livro2` | Objetos criados a partir da classe. |

```python
# Classe filha que herda da classe Livro
class LivroDigital(Livro):

    # Inicializa os atributos do livro digital
    def __init__(self, titulo, autor, preco, formato):
        super().__init__(titulo, autor, preco)  # Inicializa os atributos da classe pai
        self.formato = formato  # Adiciona um novo atributo


# Cria um objeto da classe LivroDigital
livro1 = LivroDigital("Dom Casmurro", "Machado de Assis", 29.90, "PDF")
```

```python
# Executa o método herdado da classe Livro
livro1.apresentar()

# Exibe o atributo específico da classe LivroDigital
print("Formato:", livro1.formato)

# Conceito principal: a herança permite reaproveitar atributos e
# métodos de uma classe existente, acrescentando novas características
# sem precisar reescrever toda a implementação.
```

ETL com Classes: Etapa de Transformação

```python
import pandas as pd  # Importa a biblioteca Pandas

# Define a classe responsável pela transformação
class Transformacao:

    # Recebe e armazena o DataFrame
    def __init__(self, df):
        self.df = df  # Armazena os dados no objeto

    # Método para transformar os dados
    def transformar(self):
        self.df["aprovadas"] = self.df["produzidas"] - self.df["defeituosas"]  # Calcula as peças aprovadas
        return self.df  # Retorna o DataFrame transformado
```

```python
import pandas as pd

# Cria um DataFrame de exemplo
df = pd.DataFrame({
    "produzidas": [100, 200, 150],  # Quantidade produzida
    "defeituosas": [5, 10, 3]  # Quantidade defeituosa
})

display(df)

print(40*"-")
```

```python
# Cria um objeto da classe Transformacao
etapa = Transformacao(df)

# Executa o método de transformação
df_tratado = etapa.transformar()

# Exibe o resultado
display(df_tratado)
```

| Elemento | Finalidade |
|---|---|
| `class Transformacao:` | Cria a classe responsável pelo tratamento. |
| `__init__(self, df)` | Recebe o DataFrame quando o objeto é criado. |
| `self.df = df` | Armazena o DataFrame no objeto. |
| `def transformar(self):` | Define o método que realiza o tratamento. |
| `etapa = Transformacao(df)` | Cria o objeto com os dados. |
| `etapa.transformar()` | Executa a transformação. |
