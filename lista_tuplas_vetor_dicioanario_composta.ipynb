```python
produtos["preco"]
```

```python
# forma mais segura de obetr um valor
produtos.get("estoque")
```

```python
# modificar

produtos["preco"] = 3200

produtos
```

```python
# adicionar

produtos["categoria"] = "Informatica"

produtos
```

```python
# lista dentro de um dicionário

produtos = {
       "nome":"Notebook",
       "preco": 3500,
       "cores": ["preto", "prata", "cinza"]       
}

produtos
```

```python
produtos["cores"]
```

```python
# for com dicionarios

for chave in produtos:
       print(chave)
```

```python
produtos
```

```python
# forma muito comum de percorrer dicionários: items()

for chave, valor in produtos.items():
       print(chave, valor)
```

#### ESTRUTURAS COMPOSTAS

```python
# lista de dicionários

produtos = [
       {
              "produto":"Notebook",
              "preco": 3500,
              "estoque": 8
       },
       {
              "produto":"Mouse",
              "preco": 120,
              "estoque": 30
       },
       {
              "produto":"Monitor",
              "preco": 900,
              "estoque": 12
       }

       ]
produtos
```

```python
for produto in produtos:
       print(produto["produto"])
```

```python
for produto in produtos:
       print(produto["produto"], produto["preco"])
```

```python
# dicionário com listas

vendas = {

       "produto":["Notebok", "Mouse", "Monitor"],
       "preco": [3500, 120, 900],
       "quantidade": [2, 10, 5]
}

vendas
```

```python
import pandas as pd

df = pd.DataFrame(vendas)

df
```

```python
receita = []

for preco, quantidade in zip(df["preco"], df["quantidade"]):
       receita.append(preco * quantidade)
```

```python
df["receita"] = receita

df
```
