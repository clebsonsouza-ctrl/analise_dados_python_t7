```python
# pip install selenium # Instala a biblioteca Selenium no ambiente Python

from selenium import webdriver  # Importa o WebDriver da biblioteca Selenium
from selenium.webdriver.common.by import By  # Importa a classe By para localizar elementos
from selenium.webdriver.support.ui import WebDriverWait # Importa a classe de espera explícita do Selenium
from selenium.webdriver.support import expected_conditions as EC # Configura uma espera de até 10 segundos
import pandas as pd 
import time # importa biblioteca para colocar tempo no código
```

Localizando elementos com XPath ou ID

O XPath (XML Path Language) é uma linguagem utilizada para localizar elementos em documentos HTML e

XML, considerando suas tags, atributos e posições na estrutura do documento.

O atributo id é utilizado para identificar um elemento HTML. Em uma página, seu valor deve ser único.

find_element() - permite localizar um elemento HTML na página, utilizando critérios como ID, classe ou nome da tag.

| Elemento         | Finalidade                                                   |
|------------------|--------------------------------------------------------------|
| `find_element()` | Localiza o primeiro elemento correspondente.                 |
| `By.TAG_NAME`    | Define que a busca será realizada pelo nome da tag HTML.     |
| `"h1"`           | Informa a tag que desejamos localizar.                       |

No Selenium, o primeiro elemento é aquele que aparece primeiro na estrutura do documento HTML

```python
navegador = webdriver.Chrome()  # Inicializa o Google Chrome

navegador.get("https://quotes.toscrape.com/")  # Acessa o site de treinamento

titulo = navegador.find_element(By.TAG_NAME, "h1")  # Localiza o primeiro elemento HTML h1

time.sleep(2) # espera por 2 segundos

print(titulo.text)  # Exibe o texto do elemento localizado (<h1>Quotes to Scrape</h1>)

navegador.quit()  # Encerra o navegador
```

Extraindo a primeira citação



- span é a tag HTML.


- class="text" é o atributo que identifica a classe CSS do elemento.


- Texto de uma citação é o conteúdo textual.

```python
navegador = webdriver.Chrome()  # Inicializa o navegador Google Chrome

navegador.get("https://quotes.toscrape.com/")  # Acessa o site de treinamento

#Determina que a busca será realizada por uma classe HTML.
citacao = navegador.find_element(By.CLASS_NAME, "text")  # Localiza a primeira citação pela classe text (find_element - no singular)

print(citacao.text)  # Exibe o texto da citação encontrada

time.sleep(2) # espera por 2 segundos

# navegador.quit()  # Encerra o navegador e a sessão do Selenium
```

Extraindo várias citações com Selenium

```python
navegador = webdriver.Chrome()  # Inicializa o Google Chrome

navegador.get("https://quotes.toscrape.com/")  # Acessa o site de treinamento

# armazenará uma lista de objetos WebElement.
citacoes = navegador.find_elements(By.CLASS_NAME, "text")  # Localiza todas as citações da página (find_elements - no plural)

for citacao in citacoes:  # Percorre cada elemento da lista de citações
    print(citacao.text)  # Exibe o texto da citação encontrada

time.sleep(2) # espera por 2 segundos

navegador.quit()  # Encerra o navegador e a sessão do Selenium
```

Extraindo citações e seus autores



Localiza todos os elementos HTML que possuem a classe "quote"


Estrutura básica
- Tag: div, span, small.
- Atributo: class, id, type.
- Valor do atributo: "quote", "text", "author".
- Conteúdo: texto localizado entre as tags de abertura e fechamento.

```python
# Pega os autores e citação

navegador = webdriver.Chrome()  # Inicializa o Google Chrome

navegador.get("https://quotes.toscrape.com/")  # Acessa o site de treinamento


blocos = navegador.find_elements(By.CLASS_NAME, "quote")  # Localiza todos os blocos de citações

for bloco in blocos:  # Percorre cada bloco encontrado

    # realiza a pesquisa somente dentro do elemento HTML armazenado em bloco

    citacao = bloco.find_element(By.CLASS_NAME, "text")  # Localiza a citação dentro do bloco

    autor = bloco.find_element(By.CLASS_NAME, "author")  # Localiza o autor dentro do mesmo bloco

    print("Autor:", autor.text)  # Exibe o nome do autor

    print("Citação:", citacao.text)  # Exibe o texto da citação

    print("-" * 50)  # Imprime uma linha para separar os resultados
```

Armazenando dados coletados com Selenium

```python
navegador = webdriver.Chrome()  # Inicializa o Google Chrome

navegador.get("https://quotes.toscrape.com/")  # Acessa o site de treinamento

blocos = navegador.find_elements(By.CLASS_NAME, "quote")  # Localiza todos os blocos de citações

dados = []  # Cria uma lista vazia para armazenar os dados coletados

for bloco in blocos:  # Percorre cada bloco de citação encontrado

    citacao = bloco.find_element(By.CLASS_NAME, "text")  # Localiza o texto da citação

    autor = bloco.find_element(By.CLASS_NAME, "author")  # Localiza o nome do autor

    registro = {"autor": autor.text, "citacao": citacao.text}  # Cria um dicionário com autor e citação

    dados.append(registro)  # Adiciona o dicionário à lista de dados

print("Total de registros:", len(dados))  # Exibe a quantidade de registros coletados

print(dados)  # Exibe a lista de dicionários no terminal

time.sleep(2) # espera por 2 segundos

navegador.quit()  # Encerra o navegador
```

```python
df_citacoes = pd.DataFrame(dados)

df_citacoes
```
