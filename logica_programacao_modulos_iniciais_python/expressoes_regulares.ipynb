## REGEX

| Regex      | Significado                     | Exemplo          |
| ---------- | ------------------------------- | ---------------- |
| `\d`       | Número                          | `5`              |
| `\D`       | Tudo que não é número           | `A`, `-`, espaço |
| `\w`       | Letra, número ou `_`            | `Python`, `10`   |
| `\s`       | Espaço em branco                | espaço           |
| `.`        | Praticamente qualquer caractere | `A`, `5`, `@`    |
| `+`        | Uma ou mais ocorrências         | `123`            |
| `*`        | Zero ou mais ocorrências        | —                |
| `?`        | Zero ou uma ocorrência          | —                |
| `[0-9]`    | Números de 0 a 9                | `7`              |
| `[A-Za-z]` | Letras                          | `Python`         |
| `^`        | Início do texto                 | —                |
| `$`        | Final do texto                  | —                |

```python
import re
```

```python
texto = "Pedido 4587"

resultado = re.findall(r"\d", texto)

print(resultado)
```

```python
texto = "Pedido 4587"

resultado = re.findall(r"\d+", texto)

print(resultado)
```

```python
# tudo que não é número

texto = "ABC123"

resultado = re.findall(r"\D", texto)

print(resultado)
```

```python
texto = "ABC123"

resultado = re.findall(r"\D+", texto)

print(resultado)
```

```python
telefone = "(83) 99999-1234"

telefone_limpo = re.sub(r"\D", "", telefone)

print(telefone_limpo)
```

```python
# \w => letras, números e "_"

texto = "Python_2026!"

resultado = re.findall(r"\w", texto)

print(resultado)
```

```python
texto = "Python_2026!"

resultado = re.findall(r"\w+", texto)

print(resultado)
```

```python
# \s => espaço em branco

texto = " Ana Silva "

resultado = re.findall(r"\s", texto)

resultado
```

```python
nome = "Ana    Silva"

nome_limpo = re.sub(r"\s+", " ", nome)

print(nome_limpo)
```

```python
# . => praticamente qualquer caractere

texto = "gato peixe sapo"

# qualquer caractere seguido de ato
resultado = re.findall(r".ato", texto)

print(resultado)
```

```python
texto = "gato peixe sapo"

# qualquer caractere seguido de ato
resultado = re.findall(r"gat.", texto)

print(resultado)
```

```python
# ? => zero ou uma ocorrencia

cep = "58000-123"

resultado = re.search(r"\d{5}-?\d{3}", cep)

print(resultado.group())
```

### Quantificadores Regex

- `*` → o elemento pode aparecer **zero ou várias vezes**
- `?` → o elemento pode aparecer **zero ou uma vez**
- `+` → o elemento deve aparecer **uma ou várias vezes**

#### Exemplos

- `\s*` → nenhum espaço ou vários espaços
- `-?` → hífen opcional
- `\d+` → um ou mais números

```python
# ^ → início do texto

# o padrão precisa estar no início do texto.

codigo = "PROD123"

resultado = re.search(r"^PROD", codigo)

print(resultado.group())
```

```python
# não vai, precisa estar no inicio

import re

codigo = "123PROD"

resultado = re.search(r"^PROD", codigo)

print(resultado)
```

```python
# $ → final do texto
# queremos verificar se um arquivo termina em .csv

import re

arquivo = "vendas.csv" #"vendas..........csv"

resultado = re.search(r"\.csv$", arquivo)

print(resultado.group())
```
