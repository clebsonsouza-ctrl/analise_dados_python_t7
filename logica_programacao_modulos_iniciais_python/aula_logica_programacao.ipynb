```python
#enumerate()

produtos = ["Notebook", "Mouse", "Teclado"]

for indice, produto in enumerate(produtos):
    print("posição:",indice,"-", "produto",produto)
```

```python
# for junto com if


vendas = [1500, 3200, 800, 4100, 2500]

for venda in vendas:
  if venda > 3000:
      print(venda)
```

```python
# for + if + else

vendas = [1500, 3200, 800, 4100]

for venda in vendas:

    if venda >= 3000:
        print(venda, "- Meta atingida")

    else:
        print(venda, "- Abaixo da meta")
```

```python
# “break” — interrompendo o “for”

produtos = ["Notebook", "Mouse", "Teclado", "Monitor"]

for produto in produtos:
    print(produto)

    if produto == "Teclado":
        break
```

```python
#“continue” — pule esta repetição

vendas = [1500, 0, 2300, 0, 4100]

for venda in vendas:

    if venda == 0:
        continue

    print(venda)
```

```python
# “for” dentro de outro “for"

for i in range(3): #[0, 1, 2]
    for j in range(3): #[0, 1, 2]
        print(i, j)
```

```python
# “for” dentro de outro “for"

'''
Definir lista de LOJAS = ["loja1", "loja2", "loja3"]
Definir lista de PRODUTOS = ["notebook", "celular", "impressora", "webcam"]

Para cada LOJA dentro da lista de LOJAS:
    Para cada PRODUTO dentro da lista de PRODUTOS:
        Mostrar na tela: a LOJA atual e o PRODUTO atual

    Mostrar na tela: uma linha separadora com 35 traços
'''

lojas = ["loja1", "loja2", "loja3"]

produtos = ["notebook", "celular", "impressora", "webcam"]

for loja in lojas:
  for produto in produtos:
    print("loja:", loja, "-", "produto:", produto)
  print(35*"-")
```

```python
# “for” dentro de outro “for"

lojas = ["loja1", "loja2", "loja3"]

produtos = ["notebook", "celular", "impressora", "webcam"]

cores = ["Azul", "Preto"]

for loja in lojas:
  for produto in produtos:
    for cor in cores:
      print("loja:", loja, "-", "produto:", produto, "-", "cor:", cor)
  print(45*"-")
```

```python
# Percorrendo duas listas ao mesmo tempo

produtos = ["Notebook", "Mouse", "Monitor"]
precos = [3500, 80, 1200]


for produto, preco in zip(produtos, precos):
    print("produto:", produto, "preço:", preco)
```

```python
# “while” com entrada do usuário

senha = ""

while senha != "python123":
  senha = input("DIGITE A SENHA: ")
print('Acesso autorizado!!!!')
```

```python
# Outro exemplo com entrada do usuário

producao = 0
meta = 1000

while producao < meta:

    quantidade = int(
        input("Informe a produção do ciclo: ")
    )
    producao = producao + quantidade

    print("Produção acumulada:", producao)

print("Meta atingida!")
```

```python
# while True

while True:

    resposta = input("Digite SAIR para encerrar: ")

    if resposta == "SAIR":
        break

    print("Você digitou:", resposta)
```

```python
# continue no while

numero = 0

while numero < 10:

    numero = numero + 1

    if numero == 5:
        continue

    print(numero)
```

```python
# Condições compostas no while

temperatura = 70

estoque = 5


while temperatura < 100 and estoque > 0:

    print(
        "Temperatura:", temperatura,
        "- Estoque:", estoque
    )

    temperatura = temperatura + 5
    estoque = estoque - 1
```

```python
# while...else

senha_correta = "python123"

tentativa = 1

while tentativa <= 3:

    senha = input("Digite a senha: ")

    if senha == senha_correta:
        print("Acesso autorizado.")
        break

    tentativa = tentativa + 1

else:
    print("Número máximo de tentativas atingido.")
```

```python
# while com for

maquinas = ["Prensa", "Torno", "Fresadora"]

turno = 1

while turno <= 3:

    print("\nTurno:", turno)

    for maquina in maquinas:
        print("Verificando máquina:", maquina)

    turno = turno + 1
```
