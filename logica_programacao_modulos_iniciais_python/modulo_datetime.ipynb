## MÓDULO DATE TIME

```python
# importa a classe date do módulo datetime
from datetime import date
from datetime import datetime
from datetime import timedelta
```

```python
# Obtendo a data atual
hoje = date.today()

print(hoje)
```

```python
# pegar a data separadamente cada parte

print("Dia:", hoje.day)
print("Mês:", hoje.month)
print("Ano:", hoje.year)
```

```python
# criar uma data manualmente

data_curso = date(2026, 10, 15)

print(data_curso)
```

```python
# a data e hora

agora = datetime.now()

print(agora)
```

```python
# formatando uma data

agora = datetime.now()

data_formatada = agora.strftime("%d/%m/%Y") #y 26

print(data_formatada)
```

| Código | Representa | Exemplo |
|:---:|---|:---:|
| **`%d`** | Dia | `28` |
| **`%m`** | Mês | `09` |
| **`%Y`** | Ano | `2026` |
| **`%H`** | Hora | `15` |
| **`%M`** | Minuto | `30` |
| **`%S`** | Segundo | `45` |

```python
data_hora = agora.strftime("%d/%m/%Y %H:%M")

print(data_hora)
```

```python
print(type(data_hora))

print(type(hoje))
```

```python
# Calculando a diferença entre datas

data_venda = datetime(2026, 9, 10)
data_entrega = datetime(2026, 9, 15)

diferenca = data_entrega - data_venda

print(diferenca.days)
```

```python
# adicionar dias a uma data

prazo = timedelta(days = 7)

nova_data_entrega = data_entrega + prazo

print(data_entrega)
print(30*"-")
print(nova_data_entrega)
```

```python
# Transformando um texto em data com strptime()

data = "25/09/2026"

print(type(data))

print(data)
```

```python
# Importa datetime
from datetime import datetime

# Data armazenada como texto
data_texto = "25/09/2026"

# Converte o texto para datetime
data = datetime.strptime(data_texto, "%d/%m/%Y")

# Exibe o resultado
print(data)

# Exibe o tipo
print(type(data))
```

#### Principais recursos do módulo `datetime`

- `date.today()` → Data atual
- `datetime.now()` → Data e hora atuais
- `strftime()` → Converte data para texto formatado
- `strptime()` → Converte texto para data
- `timedelta()` → Representa um intervalo de tempo
